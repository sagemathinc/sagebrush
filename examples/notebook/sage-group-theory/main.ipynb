## Permutation groups

Computed in this page by Sagebrush's own group engine, written in Rust from the literature (Sims, Holt, Seress, Atkinson). A group is stored as a *stabilizer chain*, built by the Schreier–Sims algorithm and then checked by its deterministic test, so orders and membership are proven, not probable. On all 2604 transitive groups of degree at most 16, it agrees with Magma on order, primitivity, solvability, transitivity, block systems and more. Points are $1, \dots, n$, and elements multiply left to right, as in Sage and GAP.

```python
G = PermutationGroup(['(1,2,3,4,5)', '(1,2)'])
G, G.order(), G == SymmetricGroup(5)
```

```python
# g*h applies g first
g = G('(1,2,3)(4,5)'); h = G('(1,4)')
g*h, h*g, g^-1, g.order(), g.cycle_type(), g.sign()
```

### The Mathieu groups

The five sporadic Mathieu groups are the only 4- and 5-transitive groups besides the symmetric and alternating ones. Here $M_{24}$ is built from Conway's generators on the projective line over $\mathbb{F}_{23}$, and $M_{23}$ and $M_{22}$ are its point stabilizers.

```python
[(n, MathieuGroup(n).order(), MathieuGroup(n).transitivity()) for n in [11, 12, 22, 23, 24]]
```

```python
# The number of elements of each cycle type in M11: its conjugacy class sizes
MathieuGroup(11).cycle_type_counts()
```

```python
M = MathieuGroup(12)
g = M.random_element()
g, g in M, '(1,2)' in M, M.is_primitive()
```

### Blocks and primitivity

A *block* is a set of points that every element maps to itself or to a disjoint set. The symmetries of a hexagon preserve the three pairs of opposite vertices and the two triangles. For the Galois group of a polynomial, the block systems correspond to the subfields of the field generated by one root.

```python
D = DihedralGroup(6)
D.is_primitive(), D.blocks_all(representatives=False)
```

```python
[(n, CyclicPermutationGroup(n).blocks_all()) for n in [6, 8, 12]]
```

### Solvability

$S_n$ is solvable exactly when $n \le 4$. That is why there is no formula in radicals for the roots of a general quintic. The derived series of $S_4$ is $S_4 > A_4 > V_4 > 1$.

```python
[(n, SymmetricGroup(n).is_solvable()) for n in range(2, 9)]
```

```python
[H.order() for H in SymmetricGroup(4).derived_series()]
```

```python
# PSL(2, 7), the simple group of order 168, acting on the 8 points of the projective line over F_7
P = PSL(2, 7)
P.order(), P.degree(), P.transitivity(), P.is_perfect()
```
