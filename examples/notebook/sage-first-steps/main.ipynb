```python
# The Mersenne exponents below 1000, by Lucas-Lehmer
def lucas_lehmer(p):
    s, M = 4, 2^p - 1
    for _ in range(p - 2):
        s = (s^2 - 2) % M
    return s == 0

[p for p in prime_range(3, 1000) if lucas_lehmer(p)]
```
