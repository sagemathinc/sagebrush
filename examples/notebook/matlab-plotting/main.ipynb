## Plotting gallery

The plot types a MATLAB program reaches for most, each in a few lines. Figures show when the cell ends. Line plots with styles and legends, and a 3-D surface you can turn with the mouse, are in the first MATLAB notebook; `mesh` and `plot3` draw in 3-D the same way.

```matlab
figure
subplot(2, 2, 1), bar([3 5 2 7 4]), title('bar')
subplot(2, 2, 2), stairs(0:6, [1 3 2 4 3 5 4]), title('stairs')
subplot(2, 2, 3), stem(0:10, exp(-(0:10)/3)), title('stem')
subplot(2, 2, 4), area(linspace(0, 1, 50), linspace(0, 1, 50).^2), title('area')
```

```matlab
figure
rng(5);
px = randn(60, 1); py = 0.6*px + 0.5*randn(60, 1);
scatter(px, py, 14, hypot(px, py), 'filled')
axis equal, title('scatter, colored by distance')
```

```matlab
figure
[X, Y] = meshgrid(linspace(-3, 3, 16));
Z = peaks(X, Y);
contour(X, Y, Z, 8)
title('contour of peaks'), axis equal tight
```
