## Plotting gallery

The plot types a MATLAB program reaches for most, as six subplots of one figure; figures show when the cell ends. Line plots with styles and legends, and a 3-D surface you can turn with the mouse, are in the first MATLAB notebook; `mesh` and `plot3` draw in 3-D the same way.

```matlab
figure
rng(5);
px = randn(20, 1);  py = 0.6*px + 0.5*randn(20, 1);
[X, Y] = meshgrid(linspace(-3, 3, 9));
subplot(2, 3, 1), bar([3 5 2 7 4]), title('bar')
subplot(2, 3, 2), stairs(0:6, [1 3 2 4 3 5 4]), title('stairs')
subplot(2, 3, 3), stem(0:6, exp(-(0:6)/3)), title('stem')
subplot(2, 3, 4), area(linspace(0, 1, 12), linspace(0, 1, 12).^2), title('area')
subplot(2, 3, 5), scatter(px, py, 12, hypot(px, py), 'filled'), title('scatter')
subplot(2, 3, 6), contour(X, Y, peaks(X, Y), 5), title('contour')
```
