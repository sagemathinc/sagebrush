## MATLAB

MATLAB programs, translated to Python in this page and run on Sagebrush's numpy. Values print as MATLAB prints them (checked against GNU Octave and MATLAB's documentation), variables and functions stay for the next cells, and figures show when a cell ends.

```matlab
% Matrices and linear algebra
A = [4 -2; 1 1]
x = A \ [2; 3]
[V, D] = eig([2 1; 1 3]);
diag(D)'
```

```matlab
% A function defined in a cell is there for the cells after it
function y = collatz(n)
  y = n;
  while n ~= 1
    if mod(n, 2) == 0, n = n / 2; else, n = 3 * n + 1; end
    y(end + 1) = n;
  end
end
```

```matlab
c = collatz(27);
fprintf('%d steps, peak %d\n', numel(c) - 1, max(c))
```

```matlab
% Structs and cell arrays
s = struct('name', {'Ada', 'Emmy'}, 'born', {1815, 1882});
for k = 1:numel(s)
  fprintf('%s was born in %d\n', s(k).name, s(k).born);
end
names = upper({s.name})
```

```matlab
% Figures
x = linspace(0, 2*pi, 200);
plot(x, sin(x), x, cos(x), '--')
legend('sin', 'cos'); title('Figures show when the cell ends'); grid on
```
