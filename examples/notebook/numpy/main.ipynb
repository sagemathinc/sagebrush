```python
import numpy as np
a = np.arange(12).reshape(3, 4)
print(a)
a.sum(axis=0), a.mean(), a[a % 3 == 0], a.T @ a
```

```python
# Seeded streams match NumPy's exactly
np.random.seed(0)
x = np.random.randn(100_000)
x.mean(), x.std(), np.percentile(x, [5, 50, 95])
```

```python
A = np.array([[4.0, 1, 2], [1, 3, 0], [2, 0, 5]])
w, v = np.linalg.eigh(A)
print(w)
np.linalg.solve(A, [1, 2, 3]), np.linalg.det(A), np.allclose(v @ np.diag(w) @ v.T, A)
```

```python
# Least squares and polynomials
x = np.linspace(0, 1, 50)
y = 3 * x**2 - x + 0.05 * np.sin(40 * x)
p = np.polyfit(x, y, 2)
p, np.roots(p)
```

```python
# A spectrum with the FFT
import matplotlib.pyplot as plt
t = np.linspace(0, 1, 512, endpoint=False)
signal = np.sin(2 * np.pi * 50 * t) + 0.5 * np.sin(2 * np.pi * 120 * t) + 0.3 * np.random.randn(512)
freq = np.fft.rfftfreq(512, d=1 / 512)
plt.figure(figsize=(7, 3)); plt.plot(freq, np.abs(np.fft.rfft(signal))); plt.xlabel('Hz'); plt.title('|FFT|')
```
