## Plotting gallery

The plot types a MATLAB program reaches for most, each in a few lines. Figures show when the cell ends.

```matlab
figure
x = linspace(0, 2*pi, 100);
plot(x, sin(x), '-', x, cos(x), '--', x, sin(2*x), ':', 'LineWidth', 1.5)
legend('sin x', 'cos x', 'sin 2x'), grid on
xlabel('x'), title('Line styles')
```

```matlab
figure
subplot(2, 2, 1), bar([3 5 2 7 4]), title('bar')
subplot(2, 2, 2), stairs(0:6, [1 3 2 4 3 5 4]), title('stairs')
subplot(2, 2, 3), stem(0:10, exp(-(0:10)/3)), title('stem')
subplot(2, 2, 4), area(linspace(0, 1, 50), linspace(0, 1, 50).^2), title('area')
```

```matlab
figure
rng(5);
px = randn(300, 1); py = 0.6*px + 0.5*randn(300, 1);
scatter(px, py, 12, hypot(px, py), 'filled')
colorbar, axis equal, title('scatter, colored by distance')
```

```matlab
figure
[X, Y] = meshgrid(linspace(-3, 3, 45));
Z = peaks(X, Y);
contourf(X, Y, Z, 12), colorbar
title('contourf of peaks'), axis equal tight
```

```matlab
figure
[SX, SY] = meshgrid(linspace(-3, 3, 32));
surf(SX, SY, peaks(SX, SY)), shading interp
colormap(turbo), colorbar
title('surf'), view(-30, 35)
```

```matlab
figure
[mx, my] = meshgrid(linspace(-2, 2, 25));
mesh(mx, my, mx .* exp(-mx.^2 - my.^2))
title('mesh'), view(-40, 30)
```

```matlab
figure
s = linspace(0, 8*pi, 400);
plot3(cos(s), sin(s), s/(8*pi), 'LineWidth', 1.5)
grid on, title('plot3: a helix'), view(-30, 25)
```

```matlab
figure
img = peaks(80);
subplot(1, 3, 1), imagesc(img), colormap(gca, parula), axis equal off, title('parula')
subplot(1, 3, 2), imagesc(img), colormap(gca, hot), axis equal off, title('hot')
subplot(1, 3, 3), imagesc(img), colormap(gca, gray), axis equal off, title('gray')
```
