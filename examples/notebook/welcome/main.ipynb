```python
def primes(n):
    sieve = bytearray([1]) * (n + 1)
    sieve[:2] = b'\x00\x00'
    for p in range(2, int(n**0.5) + 1):
        if sieve[p]:
            sieve[p*p::p] = bytes(len(range(p*p, n + 1, p)))
    return [i for i, ok in enumerate(sieve) if ok]

ps = primes(10**5)
print(len(ps), ps[-5:])
```
