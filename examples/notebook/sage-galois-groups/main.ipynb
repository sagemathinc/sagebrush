## Galois groups

The Galois group of an irreducible polynomial $f \in \mathbb{Q}[x]$ permutes its $n$ roots, so, up to the numbering of the roots, it is one of the *transitive groups* $n$T$k$ of degree $n = \deg f$. Sagebrush computes it for degrees $1$ to $13$, $17$, $19$ and $23$ (PARI stops at $11$), entirely in this page:

1. **Frobenius cycle types.** For a prime $p$ not dividing the discriminant, the degrees of the factors of $f$ mod $p$ are the cycle lengths of an element of the Galois group (Dedekind). With the parity of the group (is the discriminant a square?) this leaves a few candidate groups, often just one.
2. **Stauduhar's descent.** Otherwise, starting from $S_n$ or $A_n$: for each maximal subgroup $K$ that is still a candidate, an invariant $F$ of $K$ is evaluated at the roots, computed $p$-adically, permuted by coset representatives. The group lies in a conjugate of $K$ exactly when a value is a rational integer of known size, and only cosets fixed by the Frobenius at $p$ can work.

The 508 transitive groups of these degrees, and the maximal subgroups of each, were computed by Sagebrush itself; only their standard numbering and names were matched against GAP's library.

```python
R.<x> = QQ[]
for f in [x^2 + 1, x^3 - 2, x^4 - 2, x^5 - 2, x^5 - x - 1, x^6 + x^5 + x^4 + x^3 + x^2 + x + 1]:
    G = f.galois_group()
    print(f, '   ', G.transitive_label(), G.order(), G.name())
```

### How it was found

The log lists each descent step: the maximal subgroup, its index, how many cosets fixed by the Frobenius gave an integer, and the invariant used (a sum over an orbit of monomials, or a product of differences of roots or of block sums). Repeated values are separated by a *Tschirnhausen transformation* of the roots.

```python
G = (x^8 - 2).galois_group()
G.galois_log()
```

### Degrees 12 and 13

Here PARI gives up. The 13th cyclotomic polynomial has the cyclic group $C_{12}$, $x^{12} - 2$ has a group of order $48 = 12 \cdot \varphi(12)$, and compositions $g(h(x))$ have imprimitive groups, whose block systems correspond to the subfields. In prime degree $13$ every transitive group is primitive: $x^{13} - 2$ has the Frobenius group $F_{156} = \mathbb{F}_{13} \rtimes \mathbb{F}_{13}^\times$.

```python
for f in [sum(x^i for i in range(13)), x^12 - 2, x^12 + 3, (x^4 - 2)(x^3 - 3*x + 1), (x^6 - 2)(x^2 + x), x^12 - x - 1, x^13 - 2]:
    G = f.galois_group()
    print(G.transitive_label(), G.order(), G.name())
```

### The Mathieu group $M_{23}$

In August 2026, Huang, Jackson, Lee, Poonen, Pries and Zhang proved that $M_{23}$, the last of the 26 sporadic simple groups not yet known to occur, is a Galois group over $\mathbb{Q}$ ([arXiv:2608.08538](https://arxiv.org/abs/2608.08538)), with this polynomial of degree $23$. Here the cycle types of Frobenius elements rule out every group but $M_{23}$ and $A_{23}$ (rigorously), and one resolvent, the sum over the $253$ blocks of the Steiner system $S(4,7,23)$ preserved by $M_{23}$ of the products of the roots in a block, picks $M_{23}$: of the $[A_{23}:M_{23}] \approx 1.3 \cdot 10^{15}$ cosets only the few fixed by a Frobenius element need to be tried. That last step is a high-confidence computation, not a proof (the paper's proof is geometric).

### The transitive groups

There are $1, 1, 2, 5, 5, 16, 7, 50, 34, 45, 8, 301$ transitive groups of degrees $1$ to $12$ up to conjugacy (OEIS A002106). Number $295$ of degree $12$ is the Mathieu group $M_{12}$, which is $5$-transitive.

### Proofs

A step is *proven* when its value is shown to be an integer (by a norm bound, which needs high $p$-adic precision) and to differ from the values at all other cosets. By default the steps that are cheap to prove are proven. With `proof=True`, all of them are, however long that takes. Either way, the answers agree with Magma on thousands of test polynomials.

```python
G = (x^12 - 2).galois_group(proof=True)
G, G.proven
```
