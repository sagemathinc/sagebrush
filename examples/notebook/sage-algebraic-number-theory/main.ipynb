## Algebraic number theory

Number fields on Sagebrush's Rust engine, in your browser: maximal orders, prime decomposition, class groups, units and regulators, quadratic fields with 40-digit discriminants, integer factorization (ECM) and lattices. Class groups and regulators assume the Generalized Riemann Hypothesis, as Sage's and PARI's do, with a generator bound computed for each field; the results are checked against PARI. Everything is written from the literature (Cohen's books and papers) and is MIT OR Apache-2.0.

```python
# Gauss's class number one problem (Heegner, Baker, Stark): the imaginary quadratic fields with h = 1.
# h(D) counts the reduced forms a*x^2 + b*x*y + c*y^2 of discriminant D = b^2 - 4*a*c < 0:
# primitive, |b| <= a <= c, and b >= 0 when |b| = a or a = c (Gauss's reduction)
def h(D):
    n = 0
    for a in range(1, isqrt(-D // 3) + 1):
        for b in range(-a + 1, a + 1):
            c, r = divmod(b^2 - D, 4*a)
            if r == 0 and c >= a and not (c == a and b < 0) and gcd([a, b, c]) == 1:
                n += 1
    return n
[d for d in range(1, 200) if is_squarefree(d) and h(-d if d % 4 == 3 else -4*d) == 1], QuadraticField(-163).class_number()
```

```python
# A 40-digit discriminant, by a subexponential sieve
K = QuadraticField(-(10^40 + 3))
K.class_group()
```

```python
# Real quadratic: a huge class number and a tiny regulator, since 10^20 + 1 = (10^10)^2 + 1
K.<s> = QuadraticField(10^20 + 1)
K.class_number(), K.regulator()
```
