## Text, cells, structs and functions

The parts of MATLAB that aren't matrices of numbers: text, containers for mixed data, and functions as values.

### Text

A character vector (`'...'`) is a row of characters; a string (`"..."`) is one value. `sprintf` formats numbers into text, and `upper`, `strrep`, `strsplit` and `strjoin` reshape it.

```matlab
name = 'sagebrush';
greeting = sprintf('Hello, %s! pi is about %.4f.', upper(name(1)), pi)
words = strsplit('the quick brown fox', ' ')
joined = strjoin(fliplr(words), '-')
replaced = strrep('a cat and a hat', 'at', 'og')
s = "strings " + "join with +"
n = strlength(s)
```

### Regular expressions

`regexp` finds patterns: here every date written as year-month-day, with named tokens for the parts.

```matlab
log_text = 'Released 2024-03-14, patched 2024-04-02 and 2025-01-30.';
dates = regexp(log_text, '\d{4}-\d{2}-\d{2}', 'match')
parts = regexp(log_text, '(?<year>\d{4})-(?<month>\d{2})', 'names');
years = {parts.year}
masked = regexprep(log_text, '\d', '#')
```

### Cell arrays and structs

A cell array holds values of any type and size; `cellfun` applies a function to each. A struct array has named fields, and `{s.field}` collects one field from all of them.

```matlab
mixed = {42, 'text', [1 2 3], {true}};
classes = cellfun(@class, mixed, 'UniformOutput', false)
sizes = cellfun(@numel, mixed)
planets = struct('name', {'Mercury', 'Venus', 'Earth'}, 'au', {0.39, 0.72, 1.00});
for k = 1:numel(planets)
    fprintf('%-8s %.2f AU, year %.0f days\n', planets(k).name, planets(k).au, 365.25 * planets(k).au^1.5);
end
names = {planets.name}
```

### Function handles and closures

`@(x) ...` makes an anonymous function. It captures the values of the variables it uses when it is made, so changing them afterwards doesn't change it. Handles can be passed to other functions and returned from them.

```matlab
rate = 0.05;
grow = @(p, years) p * (1 + rate).^years;
rate = 0.5;
after10 = grow(1000, 10)
compose = @(f, g) @(x) f(g(x));
sin_of_square = compose(@sin, @(x) x.^2);
sin_of_square(sqrt(pi/2))
squares = arrayfun(@(k) k^2, 1:6)
```

### A function in a cell

A function defined in a cell is there for the cells after it, as a local function is in a script. This one returns two outputs. The cell after it also uses a struct as a lookup table: `s.(name)` reads or sets the field whose name is in `name`.

```matlab
function [lo, hi] = bounds_of(v)
    lo = min(v);
    hi = max(v);
end
```

```matlab
[lo, hi] = bounds_of([4 -1 9 2])
stock = struct('apples', 3, 'pears', 5);
item = 'plums';
stock.(item) = 2;
fruit = fieldnames(stock)'
total = sum(cellfun(@(f) stock.(f), fieldnames(stock)))
```
