The first code cell holds the setup: the functions this notebook uses. Its code is **hidden** (Jupyter's `source_hidden`): you see a collapsed bar, and can open it to read or change it. It still runs, like any cell.

```python
# Setup: continued fractions, exactly (integers and fractions only)
from fractions import Fraction
from math import isqrt


def cf_sqrt(n, terms=12):
    """The continued fraction of sqrt(n): a0; a1, a2, ... (periodic), by the classic integer recurrence."""
    a0 = isqrt(n)
    if a0 * a0 == n:
        return [a0]
    m, d, a, out = 0, 1, a0, [a0]
    while len(out) < terms:
        m = d * a - m
        d = (n - m * m) // d
        a = (a0 + m) // d
        out.append(a)
    return out


def cf_fraction(x, terms=8):
    """The continued fraction of a Fraction x (it ends: x is rational)."""
    out = []
    while len(out) < terms:
        a = x.numerator // x.denominator
        out.append(a)
        if x == a:
            break
        x = 1 / (x - a)
    return out


def convergents(cf):
    """p/q for each prefix of the continued fraction."""
    p0, q0, p1, q1 = 1, 0, cf[0], 1
    out = [Fraction(p1, q1)]
    for a in cf[1:]:
        p0, q0, p1, q1 = p1, q1, a * p1 + p0, a * q1 + q0
        out.append(Fraction(p1, q1))
    return out


def show(cf):
    print("[" + str(cf[0]) + "; " + ", ".join(map(str, cf[1:])) + "]")
    for c in convergents(cf):
        print(f"  {c.numerator}/{c.denominator}")
```

$\sqrt{2} = [1; 2, 2, 2, \dots]$, and its convergents $p/q$ are the best rational approximations: each solves Pell's equation $p^2 - 2q^2 = \pm 1$.

```python
show(cf_sqrt(2, 8))
```

The convergents of $\pi$ include $22/7$ and $355/113$ (Zu Chongzhi's, good to six decimals). Here from a rational approximation of $\pi$, exactly:

```python
pi_ish = Fraction(314159265358979, 100000000000000)
show(cf_fraction(pi_ish, 5))
```

The next cell checks Pell's equation for the first twenty convergents of $\sqrt{2}$. Its long output is **hidden** (Jupyter's `outputs_hidden`), since the last line says it all; open it to see every row.

```python
for c in convergents(cf_sqrt(2, 20)):
    p, q = c.numerator, c.denominator
    print(p, q, p * p - 2 * q * q)
print("all +1 or -1:", all(abs(c.numerator**2 - 2 * c.denominator**2) == 1 for c in convergents(cf_sqrt(2, 20))))
```
